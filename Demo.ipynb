```python
import os

def write_subfolder_paths_to_txt(folder_path, output_txt_path):
    """
    将指定文件夹下的所有一级子文件夹的路径写入到指定的 txt 文件中
    :param folder_path: 指定的文件夹路径
    :param output_txt_path: 输出的 txt 文件路径
    """
    try:
        # 检查指定的文件夹是否存在
        if not os.path.exists(folder_path):
            print(f"指定的文件夹 {folder_path} 不存在。")
            return

        # 获取指定文件夹下的所有一级子文件夹
        subfolders = [os.path.join(folder_path, name) for name in os.listdir(folder_path)
                      if os.path.isdir(os.path.join(folder_path, name))]

        # 将子文件夹路径写入 txt 文件
        with open(output_txt_path, 'w', encoding='utf-8') as file:
            for subfolder in subfolders:
                file.write(subfolder + '\n')

        print(f"已将 {folder_path} 下的一级子文件夹路径写入到 {output_txt_path} 文件中。")
    except Exception as e:
        print(f"处理过程中出现错误: {e}")


# 测试示例
folder_path = '/data/muxiangyu/pythonPrograms/StableAnimator/animation_data/video600_256_128'  # 替换为实际的文件夹路径
output_txt_path = 'animation_data/video_path_256_128.txt'  # 替换为实际的输出 txt 文件路径
write_subfolder_paths_to_txt(folder_path, output_txt_path)
```

```python
import shutil
import numpy as np

def select_and_copy_images(src_folder, dest_folder, num_images=20):
    """
    从源文件夹中选择指定数量的图像并复制到目标文件夹
    :param src_folder: 源文件夹路径
    :param dest_folder: 目标文件夹路径
    :param num_images: 选择的图像数量
    """
    try:
        # 检查源文件夹和目标文件夹是否存在
        if not os.path.exists(src_folder):
            print(f"源文件夹 {src_folder} 不存在。")
            return
        if not os.path.exists(dest_folder):
            os.makedirs(dest_folder)

        # 获取源文件夹中序号为0~100的图像文件
        images = [f for f in os.listdir(src_folder) if os.path.isfile(os.path.join(src_folder, f)) and f.startswith('frame_')]
        images = [f for f in images if 0 <= int(f.split('_')[1].split('.')[0]) <= 100]

        # 如果图像数量不足，直接复制所有图像
        if len(images) <= num_images:
            selected_images = images
        else:
            # 平均选取指定数量的图像
            indices = np.linspace(0, len(images) - 1, num_images, dtype=int)
            selected_images = [images[i] for i in indices]

        # 复制选中的图像到目标文件夹
        for image in selected_images:
            shutil.copy(os.path.join(src_folder, image), os.path.join(dest_folder, image))

        print(f"已从 {src_folder} 中选取 {num_images} 张图像并复制到 {dest_folder} 中。")
    except Exception as e:
        print(f"处理过程中出现错误: {e}")

# 调用函数
select_and_copy_images('/data/muxiangyu/pythonPrograms/StableAnimator/animation_data/video600_512_256/00006/clothes_white_complete', '/data/muxiangyu/pythonPrograms/StableAnimator/validation/validation_512_256/clothes_white_complete')
select_and_copy_images('/data/muxiangyu/pythonPrograms/StableAnimator/animation_data/video600_512_256/00006/centered_heads', '/data/muxiangyu/pythonPrograms/StableAnimator/validation/validation_512_256/centered_heads')
select_and_copy_images('/data/muxiangyu/pythonPrograms/StableAnimator/animation_data/video600_512_256/00006/pose_head_new', '/data/muxiangyu/pythonPrograms/StableAnimator/validation/validation_512_256/pose_head_new')
select_and_copy_images('/data/muxiangyu/pythonPrograms/StableAnimator/animation_data/video600_512_256/00006/pose_wo_head', '/data/muxiangyu/pythonPrograms/StableAnimator/validation/validation_512_256/pose_wo_head')
```

```python
from PIL import Image
import os
from tqdm import tqdm
def check_image_consistency(base_path):
    """
    检查指定路径下的子文件夹中的子文件夹内图像数量是否相等，以及图像大小是否都是 512x512
    :param base_path: 基础路径
    """
    try:
        # 获取所有子文件夹
        subfolders = [os.path.join(base_path, name) for name in os.listdir(base_path)
                      if os.path.isdir(os.path.join(base_path, name))]

        image_counts = []
        for subfolder in tqdm(subfolders):
            # 获取子文件夹中的子文件夹
            inner_subfolders = [os.path.join(subfolder, name) for name in os.listdir(subfolder)
                                if os.path.isdir(os.path.join(subfolder, name))]
            
            for inner_subfolder in inner_subfolders:
                # 获取所有图像文件
                images = [f for f in os.listdir(inner_subfolder) if os.path.isfile(os.path.join(inner_subfolder, f))]
                image_counts.append(len(images))

                # 检查图像大小
                for image in images:
                    image_path = os.path.join(inner_subfolder, image)
                    with Image.open(image_path) as img:
                        if img.size != (512, 512):
                            print(f"图像 {image_path} 的大小不是 512x512，而是 {img.size}")

        # 检查图像数量是否相等
        if len(set(image_counts)) > 1:
            print("子文件夹中的图像数量不相等:", image_counts)
        else:
            print("所有子文件夹中的图像数量相等:", image_counts[0])

    except Exception as e:
        print(f"处理过程中出现错误: {e}")

# 调用函数
check_image_consistency('/data/muxiangyu/pythonPrograms/StableAnimator/animation_data')
```

```python
from PIL import Image
import os
from tqdm import tqdm

def crop_images_to_512x256(src_base_path, dest_base_path):
    """
    将源路径下的子文件夹中的二级子文件夹内的图像从 512x512 剪切为 512x256，并保存到目标路径下
    :param src_base_path: 源基础路径
    :param dest_base_path: 目标基础路径
    """
    try:
        # 获取所有子文件夹
        subfolders = [os.path.join(src_base_path, name) for name in os.listdir(src_base_path)
                      if os.path.isdir(os.path.join(src_base_path, name))]

        for subfolder in tqdm(subfolders):
            # 获取子文件夹中的二级子文件夹
            inner_subfolders = [os.path.join(subfolder, name) for name in os.listdir(subfolder)
                                if os.path.isdir(os.path.join(subfolder, name))]
            
            for inner_subfolder in inner_subfolders:
                # 获取所有图像文件
                images = [f for f in os.listdir(inner_subfolder) if os.path.isfile(os.path.join(inner_subfolder, f))]
                
                # 创建目标文件夹
                relative_path = os.path.relpath(inner_subfolder, src_base_path)
                dest_folder = os.path.join(dest_base_path, relative_path)
                os.makedirs(dest_folder, exist_ok=True)

                for image in images:
                    src_image_path = os.path.join(inner_subfolder, image)
                    dest_image_path = os.path.join(dest_folder, image)

                    # 打开图像并裁剪
                    with Image.open(src_image_path) as img:
                        if img.size == (512, 512):
                            cropped_img = img.crop((128, 0, 384, 512))  # 保留中间部分
                            cropped_img.save(dest_image_path)
                        else:
                            print(f"图像 {src_image_path} 的大小不是 512x512，跳过裁剪。")

        print(f"已完成将 {src_base_path} 下的图像裁剪并保存到 {dest_base_path}。")
    except Exception as e:
        print(f"处理过程中出现错误: {e}")

# 调用函数
src_base_path = '/data/muxiangyu/pythonPrograms/StableAnimator/animation_data'
dest_base_path = '/data/muxiangyu/pythonPrograms/StableAnimator/animation_data_crop'
crop_images_to_512x256(src_base_path, dest_base_path)
```

```python
from PIL import Image
import os
from tqdm import tqdm

def downsample_images(src_base_path, dest_base_path):
    """
    将源路径下的子文件夹中的二级子文件夹内的图像从 512x256 下采样到 256x128，并保存到目标路径下
    :param src_base_path: 源基础路径
    :param dest_base_path: 目标基础路径
    """
    try:
        # 获取所有子文件夹
        subfolders = [os.path.join(src_base_path, name) for name in os.listdir(src_base_path)
                      if os.path.isdir(os.path.join(src_base_path, name))]

        for subfolder in tqdm(subfolders, desc="Processing subfolders"):
            # 获取子文件夹中的二级子文件夹
            inner_subfolders = [os.path.join(subfolder, name) for name in os.listdir(subfolder)
                                if os.path.isdir(os.path.join(subfolder, name))]
            
            for inner_subfolder in inner_subfolders:
                # 获取所有图像文件
                images = [f for f in os.listdir(inner_subfolder) if os.path.isfile(os.path.join(inner_subfolder, f))]
                
                # 创建目标文件夹
                relative_path = os.path.relpath(inner_subfolder, src_base_path)
                dest_folder = os.path.join(dest_base_path, relative_path)
                os.makedirs(dest_folder, exist_ok=True)

                for image in images:
                    src_image_path = os.path.join(inner_subfolder, image)
                    dest_image_path = os.path.join(dest_folder, image)

                    # 打开图像并下采样
                    with Image.open(src_image_path) as img:
                        if img.size == (512, 256):
                            downsampled_img = img.resize((256, 128), Image.ANTIALIAS)
                            downsampled_img.save(dest_image_path)
                        else:
                            print(f"图像 {src_image_path} 的大小不是 512x256，跳过下采样。")

        print(f"已完成将 {src_base_path} 下的图像下采样并保存到 {dest_base_path}。")
    except Exception as e:
        print(f"处理过程中出现错误: {e}")

# 调用函数、
base_path = '/data/muxiangyu/pythonPrograms/StableAnimator/animation_data'  # 替换为实际的基础路径
new_folder_name = 'video_512_256'
src_base_path = os.path.join(base_path, new_folder_name)
dest_base_path = os.path.join(base_path, 'video600_256_128')
downsample_images(src_base_path, dest_base_path)
```

```python
import os

def rename_images_with_zero_padding(folder_path):
    """
    将文件夹中的图像重命名为带有零填充的格式，例如 frame_1.png -> frame_001.png
    :param folder_path: 文件夹路径
    """
    try:
        # 获取文件夹中的所有图像文件
        images = [f for f in os.listdir(folder_path) if f.startswith('frame_') and f.endswith('.png')]

        for image in images:
            # 提取文件名中的数字部分
            number = int(image.split('_')[1].split('.')[0])
            # 生成新的文件名
            new_name = f"frame_{number:03d}.png"
            # 重命名文件
            os.rename(os.path.join(folder_path, image), os.path.join(folder_path, new_name))

        print(f"已完成将 {folder_path} 中的图像重命名为带有零填充的格式。")
    except Exception as e:
        print(f"处理过程中出现错误: {e}")

# 示例调用
folder1 = '/data/muxiangyu/pythonPrograms/StableAnimator/validation/validation_256_128/pose_head_new'
folder2 = '/data/muxiangyu/pythonPrograms/StableAnimator/validation/validation_256_128/pose_wo_head'
rename_images_with_zero_padding(folder1)
rename_images_with_zero_padding(folder2)
```

```python
from PIL import Image
import os

def concatenate_images_with_given_images(folder1, folder2, image1_path, output_folder):
    """
    将两个文件夹中名字相同的图像分别与给定的两张图像横向拼接，并保存到指定文件夹中。
    拼接的图像名称与文件夹中的图片名称相同。
    
    :param folder1: 第一个文件夹路径
    :param folder2: 第二个文件夹路径
    :param image1_path: 第一张给定图像的路径
    :param image2_path: 第二张给定图像的路径
    :param output_folder: 输出文件夹路径
    """
    try:
        # 检查输出文件夹是否存在，不存在则创建
        os.makedirs(output_folder, exist_ok=True)

        # 获取两个文件夹中名字相同的图像
        images1 = set(os.listdir(folder1))
        images2 = set(os.listdir(folder2))
        common_images = images1.intersection(images2)

        # 打开给定的两张图像
        given_image1 = Image.open(image1_path)
       

        # 拼接两个文件夹中名字相同的图像与给定的两张图像
        for image_name in common_images:
            image1 = Image.open(os.path.join(folder1, image_name))
            image2 = Image.open(os.path.join(folder2, image_name))
         

            # 创建拼接后的图像（四张图像横向拼接）
            concatenated_image = Image.new('RGB', (image1.width + image2.width + given_image1.width, image1.height))
            concatenated_image.paste(image1, (0, 0))
            concatenated_image.paste(image2, (image1.width, 0))
            concatenated_image.paste(given_image1, (image1.width + image2.width, 0))
            # concatenated_image.paste(given_image2, (image1.width + image2.width + given_image1.width, 0))

            # 保存拼接后的图像，名称与原图像名称相同
            concatenated_image.save(os.path.join(output_folder, image_name))

        print(f"图像已成功拼接并保存到 {output_folder} 中。")
    except Exception as e:
        print(f"处理过程中出现错误: {e}")

# 示例调用
folder1 = '/data/muxiangyu/pythonPrograms/StableAnimator/validation/validation_512_256/clothes_white_complete'
folder2 = '/data/muxiangyu/pythonPrograms/StableAnimator/validation/validation_512_256/pose_wo_head'
image1_path = '/data/muxiangyu/pythonPrograms/StableAnimator/validation/validation_512_256/centered_heads/frame_1.png'
output_folder = '/data/muxiangyu/pythonPrograms/StableAnimator/validation/validation_512_256/Pinjie'
concatenate_images_with_given_images(folder1, folder2, image1_path, output_folder)
```

```python
import os

def rename_images_sequentially(folder_path):
    """
    对文件夹中的图像按照顺序从小到大排序后，重命名为 frame_01, frame_02 等形式。
    :param folder_path: 文件夹路径
    """
    try:
        # 获取文件夹中的所有图像文件，并按文件名排序
        images = [f for f in os.listdir(folder_path) if f.startswith('frame_') and f.endswith('.png')]
        images.sort()

        # 按顺序重命名图像
        for index, image in enumerate(images, start=0):
            new_name = f"frame_{index:02d}.png"
            os.rename(os.path.join(folder_path, image), os.path.join(folder_path, new_name))

        print(f"已完成将 {folder_path} 中的图像重命名为连续的形式。")
    except Exception as e:
        print(f"处理过程中出现错误: {e}")

# 示例调用
folder1 = r"/data/muxiangyu/pythonPrograms/StableAnimator/validation/validation_512_256/Pinjie"
rename_images_sequentially(folder1)
```

```python
from PIL import Image
import os

def split_gif_to_frames(gif_path, output_folder):
    """
    将 GIF 文件分解成图像帧，并保存在指定文件夹中
    :param gif_path: GIF 文件路径
    :param output_folder: 输出文件夹路径
    """
    try:
        # 检查输出文件夹是否存在，不存在则创建
        os.makedirs(output_folder, exist_ok=True)

        # 打开 GIF 文件
        with Image.open(gif_path) as gif:
            frame_index = 0
            while True:
                # 保存当前帧
                frame_path = os.path.join(output_folder, f"frame_{frame_index:02d}.png")
                gif.save(frame_path, format="PNG")
                frame_index += 1

                # 尝试移动到下一帧
                try:
                    gif.seek(frame_index)
                except EOFError:
                    # 到达 GIF 的最后一帧
                    break

        print(f"GIF 文件已分解为 {frame_index} 帧，并保存在 {output_folder} 中。")
    except Exception as e:
        print(f"处理过程中出现错误: {e}")

# 示例调用
gif_path = '/data/muxiangyu/pythonPrograms/StableAnimator/checkpoints/Animation_600video_clothes_v14/validation_images/step_45000_val_img_0.gif'  # 替换为实际的 GIF 文件路径
output_folder = '/data/muxiangyu/pythonPrograms/StableAnimator/validation/validation_512_256/generatedImage'
split_gif_to_frames(gif_path, output_folder)
```

```python
from PIL import Image
import os

def concatenate_images(folder1, folder2, output_folder):
    """
    将两个文件夹中名字相同的图像横向拼接，并保存到指定文件夹中。
    
    :param folder1: 第一个文件夹路径
    :param folder2: 第二个文件夹路径
    :param output_folder: 输出文件夹路径
    """
    try:
        # 检查输出文件夹是否存在，不存在则创建
        os.makedirs(output_folder, exist_ok=True)

        # 获取两个文件夹中名字相同的图像
        images1 = set(os.listdir(folder1))
        images2 = set(os.listdir(folder2))
        common_images = images1.intersection(images2)

        # 拼接两个文件夹中名字相同的图像
        for image_name in common_images:
            image1 = Image.open(os.path.join(folder1, image_name))
            image2 = Image.open(os.path.join(folder2, image_name))

            # 创建拼接后的图像（横向拼接）
            concatenated_image = Image.new('RGB', (image1.width + image2.width, image1.height))
            concatenated_image.paste(image1, (0, 0))
            concatenated_image.paste(image2, (image1.width, 0))

            # 保存拼接后的图像
            concatenated_image.save(os.path.join(output_folder, image_name))

        print(f"图像已成功拼接并保存到 {output_folder} 中。")
    except Exception as e:
        print(f"处理过程中出现错误: {e}")

# 调用函数
folder1 = '/data/muxiangyu/pythonPrograms/StableAnimator/validation/validation_512_256/Pinjie'
folder2 = '/data/muxiangyu/pythonPrograms/StableAnimator/validation/validation_512_256/generatedImage'
output_folder = '/data/muxiangyu/pythonPrograms/StableAnimator/validation/validation_512_256/concatenated_images'
concatenate_images(folder1, folder2, output_folder)
```

```python
import cv2
import os

def images_to_video(image_folder, output_video_path, fps=10):
    """
    将给定文件夹中的图片合成为一个 MP4 格式的视频
    :param image_folder: 图片所在文件夹路径
    :param output_video_path: 输出视频文件路径
    :param fps: 视频帧率
    """
    try:
        # 获取文件夹中的所有图片文件，并按文件名排序
        images = [img for img in os.listdir(image_folder) if img.endswith(('.png', '.jpg', '.jpeg'))]
        images.sort()

        if not images:
            print(f"文件夹 {image_folder} 中没有找到图片文件。")
            return

        # 获取第一张图片的尺寸
        first_image_path = os.path.join(image_folder, images[0])
        first_image = cv2.imread(first_image_path)
        height, width, _ = first_image.shape

        # 初始化视频写入器
        fourcc = cv2.VideoWriter_fourcc(*'mp4v')  # 使用 MP4 编码
        video_writer = cv2.VideoWriter(output_video_path, fourcc, fps, (width, height))

        # 将每张图片写入视频
        for image_name in images:
            image_path = os.path.join(image_folder, image_name)
            frame = cv2.imread(image_path)
            video_writer.write(frame)

        # 释放视频写入器
        video_writer.release()
        print(f"视频已成功保存到 {output_video_path}")
    except Exception as e:
        print(f"处理过程中出现错误: {e}")

# 示例调用
image_folder =  '/data/muxiangyu/pythonPrograms/StableAnimator/validation/validation_512_256/concatenated_images' # 替换为实际的图片文件夹路径
output_video_path = '/data/muxiangyu/pythonPrograms/StableAnimator/validation/validation_512_256/output_video.mp4'  # 替换为实际的输出视频路径
images_to_video(image_folder, output_video_path)
```

```python
from animation.modules.face_model import FaceModel
import cv2
import numpy as np
from scipy.spatial.distance import cosine
face_model = FaceModel()
app=face_model.app
face_helper=face_model.face_helper
handler_ante=face_model.handler_ante

face_helper.clean_all()
validation_image_path_whole = '/data/muxiangyu/pythonPrograms/StableAnimator/frame_0012.png'
validation_image_path_crop = '/data/muxiangyu/pythonPrograms/StableAnimator/animation_data/video600_512_256/00001/centered_heads/frame_0.png'
validation_face_whole = cv2.imread(validation_image_path_whole)
validation_face_crop = cv2.imread(validation_image_path_crop)

print(type(validation_face_whole))
print(validation_face_whole.shape)

validation_image_face_whole = app.get(validation_face_whole)
validation_image_face_crop = app.get(validation_face_crop)

if len(validation_image_face_whole) > 0:
    validation_image_face_whole = sorted(validation_image_face_whole,
        key=lambda x: (x['bbox'][2] - x['bbox'][0]) * (x['bbox'][3] - x['bbox'][1]))[-1]
    validation_image_id_whole_embedding = validation_image_face_whole['embedding']
else:
        validation_image_id_whole_embedding = None
print(validation_image_id_whole_embedding)

if len(validation_image_face_crop) > 0:
    validation_image_face_crop = sorted(validation_image_face_crop,
        key=lambda x: (x['bbox'][2] - x['bbox'][0]) * (x['bbox'][3] - x['bbox'][1]))[-1]
    validation_image_id_crop_embedding = validation_image_face_crop['embedding']
else:
        validation_image_id_crop_embedding = None








def calculate_cosine_similarity(embedding1, embedding2):
    """
    Calculate cosine similarity between two embeddings.
    
    :param embedding1: First face embedding (numpy array)
    :param embedding2: Second face embedding (numpy array)
    :return: Cosine similarity score (higher means more similar, range from -1 to 1)
    """
    # Cosine distance is 1 - cosine similarity, so we subtract from 1
    similarity = 1 - cosine(embedding1, embedding2)
    return similarity


if validation_image_id_whole_embedding is not None and validation_image_id_crop_embedding is not None:
    # Calculate similarity
    similarity_score = calculate_cosine_similarity(
        validation_image_id_whole_embedding, 
        validation_image_id_crop_embedding
    )
    
    print(f"Cosine similarity between the face embeddings: {similarity_score:.4f}")

# print(validation_image_id_whole_embedding)
# print(validation_image_id_crop_embedding)


```
